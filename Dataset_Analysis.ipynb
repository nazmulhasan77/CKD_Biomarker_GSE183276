# Visualize

```python
import matplotlib.pyplot as plt

adata.obs['condition.l1'].value_counts().plot(kind='bar')
plt.title("Cells per Condition (CKD, AKI, Ref)")
plt.show()
```

```python
adata.obs['class'].value_counts().plot(kind='bar')
plt.title("Cell Class Distribution")
plt.show()
```

```python
adata.obs['subclass.l1'].value_counts().head(15).plot(kind='bar')
plt.title("Top Subclass Types")
plt.show()
```

```python
import pandas as pd

ct = pd.crosstab(adata.obs['class'], adata.obs['condition.l1'])
ct.plot(kind='bar', stacked=True)
plt.title("Cell Type vs Condition")
plt.show()
```

## UMAP

```python
adata_sub = adata[:3000].to_memory()
```

```python
# Compute UMAP if not already present
sc.pp.neighbors(adata_sub, n_neighbors=15, n_pcs=40)
sc.tl.umap(adata_sub)

```

```python
sc.pl.umap(adata_sub, color='condition.l1')
sc.pl.umap(adata_sub, color='class')
sc.pl.umap(adata_sub, color='subclass.l1')
sc.pl.umap(adata_sub, color='Phase')
sc.pl.umap(adata_sub, color=['degen.score', 'aEpi.score'])
sc.pl.umap(adata_sub, color=['collagen.score', 'matrisome.score'])
```

```python
sc.pl.violin(adata_sub, ['nCount_RNA', 'nFeature_RNA'], groupby='condition.l1')
```

```python
sc.pl.violin(adata_sub, ['percent.mt'], groupby='condition.l1')
```
